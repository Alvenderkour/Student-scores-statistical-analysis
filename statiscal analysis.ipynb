# Student Scores: Statistical Analysis

## 1. Project Objective

The objective of this project is to analyze student performance in Mathematics and Language and investigate whether scores differ by gender and parental education level using descriptive statistics and hypothesis testing.

## 2. Introduction

Statistical analysis helps us identify patterns in data and determine whether observed differences are statistically significant. In this project, student Mathematics and Language scores are analyzed to explore gender differences and differences across parental education levels.

Descriptive statistics and visualizations are used to explore the dataset. Welch's independent-samples t-tests are applied to compare scores between male and female students. One-way ANOVA is used to examine differences across parental education groups, followed by Tukey's HSD post-hoc tests to identify which specific groups differ significantly.

#### Is There Any Statistically Significant differnce between male and female students in their Mathematics  score and in their Language scores?

```python
from scipy.stats import ttest_ind

# Separate Mathematics scores by gender
female_math = df[
    (df["gender"] == "female") &
    (df["subject"] == "math")
]["score"]

male_math = df[
    (df["gender"] == "male") &
    (df["subject"] == "math")
]["score"]

# Welch's independent-samples t-test
t_stat, p_value_math = ttest_ind(
    female_math,
    male_math,
    equal_var=False
)

print("Female Mathematics mean:", round(female_math.mean(), 2))
print("Male Mathematics mean:", round(male_math.mean(), 2))
print("t-statistic:", round(t_stat, 4))
print("p-value:", p_value_math)

alpha = 0.05

if p_value_math < alpha:
    print("Result: Reject the null hypothesis.")
    print("There is a statistically significant difference in Mathematics scores.")
else:
    print("Result: Fail to reject the null hypothesis.")
    print("There is insufficient evidence of a statistically significant difference.")
```

```python
from scipy.stats import ttest_ind

# Separate language scores by gender
female_lang = df[
    (df["gender"] == "female") &
    (df["subject"] == "language")
]["score"]

male_lang= df[
    (df["gender"] == "male") &
    (df["subject"] == "language")
]["score"]

# Welch's independent-samples t-test
t_stat, p_value = ttest_ind(
    female_lang,
    male_lang,
    equal_var=False
)

print("Female language mean:", round(female_lang.mean(), 2))
print("Male language mean:", round(male_lang.mean(), 2))
print("t-statistic:", round(t_stat, 4))
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Result: Reject the null hypothesis.")
    print("There is a statistically significant difference in language scores.")
else:
    print("Result: Fail to reject the null hypothesis.")
    print("There is insufficient evidence of a statistically significant difference.")
```

### Findings: Gender Differences
The Welch's independent-samples t-tests revealed statistically significant gender differences in both subjects. In Mathematics, male students achieved a higher mean score (68.95) than female students (63.86), with a statistically significant difference (t = −3.7782, p < 0.001). In Language, female students achieved a higher mean score (73.00) than male students (64.93), also showing a statistically significant difference (t = 6.4096, p < 0.001). Therefore, the null hypothesis of equal mean scores between genders was rejected for both subjects.

```python
import pandas as pd
from scipy.stats import shapiro, levene

subjects = ["math", "language"]

education_levels = df["parental.level.of.education"].unique()

for subject in subjects:
    print(f"\n{'=' * 50}")
    print(f"ASSUMPTION CHECKS FOR {subject.upper()}")
    print(f"{'=' * 50}")

    # Select scores for the current subject
    subject_data = df[df["subject"] == subject]

    # 1. Normality test for each parental education group
    print("\n1. Shapiro-Wilk Normality Test")

    for education in education_levels:
        scores = subject_data[
            subject_data["parental.level.of.education"] == education
        ]["score"]

        statistic, p_value = shapiro(scores)

        print(f"\n{education}")
        print(f"Shapiro-Wilk statistic: {statistic:.4f}")
        print(f"p-value: {p_value:.4f}")

        if p_value > 0.05:
            print("No significant evidence against normality.")
        else:
            print("Evidence against normality.")

    # 2. Homogeneity of variance test
    groups = [
        subject_data[
            subject_data["parental.level.of.education"] == education
        ]["score"]
        for education in education_levels
    ]

    statistic, p_value = levene(*groups, center="median")

    print("\n2. Levene's Test for Equal Variances")
    print(f"Test statistic: {statistic:.4f}")
    print(f"p-value: {p_value:.4f}")

    if p_value > 0.05:
        print("No significant evidence of unequal variances.")
    else:
        print("Evidence of unequal variances.")
```

```python
from scipy.stats import f_oneway

# Separate Mathematics scores by parental education
associate_math = df[(df["subject"] == "math") &
                    (df["parental.level.of.education"] == "associate's degree")]["score"]

bachelor_math = df[(df["subject"] == "math") &
                   (df["parental.level.of.education"] == "bachelor's degree")]["score"]

highschool_math = df[(df["subject"] == "math") &
                     (df["parental.level.of.education"] == "high school")]["score"]

master_math = df[(df["subject"] == "math") &
                 (df["parental.level.of.education"] == "master's degree")]["score"]

# Perform ANOVA
f_stat, p_value = f_oneway(
    associate_math,
    bachelor_math,
    highschool_math,
    master_math
)

print("F-statistic:", round(f_stat, 4))
print("p-value:", p_value)

# Decision
if p_value < 0.05:
    print("Reject the null hypothesis.")
    print("There is a significant difference in Mathematics scores across parental education levels.")
else:
    print("Fail to reject the null hypothesis.")
    print("There is insufficient evidence of a significant difference.")
```

```python
from scipy.stats import f_oneway

# Separate Language scores by parental education
associate_language = df[(df["subject"] == "language") &
                        (df["parental.level.of.education"] == "associate's degree")]["score"]

bachelor_language = df[(df["subject"] == "language") &
                       (df["parental.level.of.education"] == "bachelor's degree")]["score"]

highschool_language = df[(df["subject"] == "language") &
                         (df["parental.level.of.education"] == "high school")]["score"]

master_language = df[(df["subject"] == "language") &
                     (df["parental.level.of.education"] == "master's degree")]["score"]

# Perform ANOVA
f_stat, p_value = f_oneway(
    associate_language,
    bachelor_language,
    highschool_language,
    master_language
)

print("F-statistic:", round(f_stat, 4))
print("p-value:", p_value)

# Decision
if p_value < 0.05:
    print("Reject the null hypothesis.")
    print("There is a significant difference in Language scores across parental education levels.")
else:
    print("Fail to reject the null hypothesis.")
    print("There is insufficient evidence of a significant difference.")
```

### Findings: Parental Education Levels

The descriptive analysis showed differences in average Mathematics and Language scores across parental education groups. Students whose parents held a master's degree achieved the highest mean scores in both Language (75.53) and Mathematics (69.75), while students whose parents had a high school education achieved the lowest mean scores in Language (63.53) and Mathematics (62.24).

One-way ANOVA showed statistically significant differences in scores across parental education levels for both subjects (p < 0.05). Therefore, the null hypotheses of equal group means were rejected for Mathematics and Language.

           ==============================================================

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Mathematics post-hoc test
math_data = df[df["subject"] == "math"]

tukey_math = pairwise_tukeyhsd(
    endog=math_data["score"],
    groups=math_data["parental.level.of.education"],
    alpha=0.05
)

print("POST-HOC RESULTS: MATHEMATICS")
print(tukey_math)


# Language post-hoc test
language_data = df[df["subject"] == "language"]

tukey_language = pairwise_tukeyhsd(
    endog=language_data["score"],
    groups=language_data["parental.level.of.education"],
    alpha=0.05
)

print("\nPOST-HOC RESULTS: LANGUAGE")
print(tukey_language)
```

### Findings: Tukey's HSD Post-hoc Analysis

Tukey's HSD post-hoc tests were conducted to identify which parental education groups differed significantly.

For Mathematics, significant differences were found between the high school group and the associate's degree group (p = 0.0037), bachelor's degree group (p = 0.0079), and master's degree group (p = 0.0055). The other pairwise comparisons were not statistically significant.

For Language, significant differences were found between the associate's degree and high school groups (p = 0.0001), associate's degree and master's degree groups (p = 0.0429), bachelor's degree and high school groups (p < 0.001), and high school and master's degree groups (p < 0.001). The remaining pairwise comparisons were not statistically significant.
